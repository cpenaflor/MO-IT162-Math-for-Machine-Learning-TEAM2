```python
#Customer Demographic EDA

import pandas as pd
import matplotlib.pyplot as plt

file_path = r"C:\Users\chang\Desktop\ML\MO-IT162-Math-for-Machine-Learning-TEAM2\src\data\processed_data_for_submission\CLEANED_customer_demographics_v3.csv"
df = pd.read_csv(file_path)


plt.figure(figsize=(6,4))
df['Gender'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title("Customer Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()


plt.figure(figsize=(6,4))
df['IncomeLevel'].value_counts().plot(kind='bar', color='lightgreen', edgecolor='black')
plt.title("Customer Count by Income Level")
plt.xlabel("Income Level")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()


plt.figure(figsize=(10,5))
df['Location'].value_counts().head(10).plot(kind='bar', color='salmon', edgecolor='black')
plt.title("Top 10 Customer Locations")
plt.xlabel("Location")
plt.ylabel("Count")
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# Customer Transaction

import pandas as pd
import matplotlib.pyplot as plt

file_path = r"C:\Users\chang\Desktop\ML\MO-IT162-Math-for-Machine-Learning-TEAM2\src\data\processed_data_for_submission\CLEANED_customer_transactions_v3.csv"
df = pd.read_csv(file_path)

# -------------------------------
# Bar chart: Product Category
# -------------------------------
plt.figure(figsize=(8,5))
df['ProductCategory'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title("Number of Transactions by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Count")
plt.xticks(rotation=45, ha='right')
plt.show()

# -------------------------------
# Bar chart: Payment Method
# -------------------------------
plt.figure(figsize=(6,4))
df['PaymentMethod'].value_counts().plot(kind='bar', color='lightgreen', edgecolor='black')
plt.title("Number of Transactions by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()

# -------------------------------
# Bar chart: High Spenders vs Non-High Spenders
# -------------------------------
plt.figure(figsize=(5,4))
df['IsHighSpender'].value_counts().plot(kind='bar', color='salmon', edgecolor='black')
plt.title("High Spenders vs Non-High Spenders")
plt.xlabel("Is High Spender (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```
