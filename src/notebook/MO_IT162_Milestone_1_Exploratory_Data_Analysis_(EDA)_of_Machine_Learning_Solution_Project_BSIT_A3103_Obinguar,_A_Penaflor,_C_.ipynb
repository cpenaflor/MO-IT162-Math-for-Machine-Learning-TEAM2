### Validate the Date Formatting

```python
def to_datetime_ymd(df, col):
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")
        print(f"{col}: parsed to datetime; invalid (NaT) count =", df[col].isna().sum())

to_datetime_ymd(demographics, "SignupDate")
to_datetime_ymd(transactions, "TransactionDate")
to_datetime_ymd(socials, "InteractionDate")

# Show date ranges
for name, df in [("demographics", demographics), ("transactions", transactions), ("social", socials)]:
    for c in df.select_dtypes("datetime").columns:
        print(f"{name}.{c} range:", df[c].min(), "→", df[c].max())
```

### Data Quality Inspection

```python
def missingness(df, name):
    miss = df.isna().sum().sort_values(ascending=False)
    ratio = (miss / len(df)).round(4)
    out = pd.DataFrame({"missing": miss, "missing_pct": ratio})
    print(f"\nMissingness for {name}:")
    display(out[out["missing"] > 0])
    return out

miss_demo = missingness(demographics, "demographics")
miss_tx   = missingness(transactions, "transactions")
miss_sm   = missingness(socials, "social")
```

### Outlier Checks

```python
def iqr_outlier_report(series, label=""):
    s = series.dropna()
    if s.empty:
        print(f"{label}: no data")
        return
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr
    outliers = s[(s < lower) | (s > upper)]
    print(f"{label} IQR bounds: [{lower:.2f}, {upper:.2f}] — Outliers: {len(outliers)}")

if "Age" in demographics.columns:
    iqr_outlier_report(demographics["Age"], "demographics.Age")

if "Amount" in transactions.columns:
    iqr_outlier_report(transactions["Amount"], "transactions.Amount")
```

#### Customer Segmentation (RFM-lite)
- Use Recency, Frequency, and Monetary value to analyze customer behavior.  
- A scatterplot can reveal clusters of loyal vs. at-risk customers.  
- Supports targeted marketing and retention strategies.  

```python
transactions["TransactionDate"] = pd.to_datetime(transactions["TransactionDate"], errors="coerce")
```

```python
cust_key = "CustomerID"
assert cust_key in transactions.columns and cust_key in demographics.columns

txg = (transactions
       .groupby(cust_key, as_index=False)
       .agg(frequency=("TransactionDate", "count"),
            monetary=("Amount", "sum"),
            last_purchase=("TransactionDate", "max")))
```

```python
ref_date = transactions["TransactionDate"].max()
txg["recency_days"] = (ref_date - txg["last_purchase"]).dt.days

# Merge with demographics (optional but useful later)
cust = demographics.merge(txg, on=cust_key, how="left")

# Safety: fill NAs for new RFM columns (e.g., customers with no tx)
cust[["frequency","monetary","recency_days"]] = cust[["frequency","monetary","recency_days"]].fillna(0)
cust.head()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x="frequency", y="monetary",
    size="recency_days", hue="recency_days",
    data=cust, alpha=0.7
)
plt.title("Customer Segmentation: RFM-lite (CustomerID)")
plt.xlabel("Frequency (Purchases)")
plt.ylabel("Monetary (Total Amount)")
plt.legend(title="Recency (days)", bbox_to_anchor=(1.03, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

- Customers with **higher frequency and higher monetary values** are rare but exist, representing the “best” customers.  
- Many customers only purchased once or twice, contributing lower revenue.  
- Recency patterns show that some customers haven’t purchased for a long time, signaling potential churn risk.

- **Data Quality**: Most variables were clean and usable after preprocessing, with missing values handled through imputation or categorical tagging (e.g., “Unknown”). Dates were standardized to `YYYY-MM-DD` format for consistency.
