## Data Cleaning

```python
print("Null values per dataset:")
print("Demographics:\n", demographics.isnull().sum())
print("Transactions:\n", transactions.isnull().sum())
print("Social Media:\n", social.isnull().sum())
```

```python
# Convert Age to numeric before imputation
demographics["Age"] = pd.to_numeric(demographics["Age"], errors="coerce")
demographics["Age"] = demographics["Age"].fillna(demographics["Age"].median())
```

```python
# Convert Amount to numeric, coerce errors if any
transactions["Amount"] = pd.to_numeric(transactions["Amount"], errors="coerce")
transactions["Amount"] = transactions["Amount"].fillna(transactions["Amount"].median())
```

```python
# Fill missing values in demographics
demographics["Age"] = demographics["Age"].fillna(demographics["Age"].median())
demographics["IncomeLevel"] = demographics["IncomeLevel"].fillna("Unknown")

# Fill missing values in transactions
transactions["Amount"] = transactions["Amount"].fillna(transactions["Amount"].median())
transactions["ProductCategory"] = transactions["ProductCategory"].fillna("Unknown")

# Fill missing values in social media
social["Platform"] = social["Platform"].fillna("Unknown")
social["Sentiment"] = social["Sentiment"].fillna("Unknown")

# Double-check if nulls remain
print("Remaining nulls after imputation:")
print("Demographics:\n", demographics.isnull().sum())
print("Transactions:\n", transactions.isnull().sum())
print("Social Media:\n", social.isnull().sum())
```

```python
# Remove unrealistic ages
demographics.loc[(demographics["Age"] < 18) | (demographics["Age"] > 100), "Age"] = np.nan

# Fill again with median
demographics["Age"] = demographics["Age"].fillna(demographics["Age"].median())

# Check after cleaning
print(demographics["Age"].describe())
```

## Date Validation

```python
demographics_v2["SignupDate"]   = pd.to_datetime(demographics_v2["SignupDate"], errors="coerce")
transactions_v2["TransactionDate"] = pd.to_datetime(transactions_v2["TransactionDate"], errors="coerce")
social_v2["InteractionDate"]   = pd.to_datetime(social_v2["InteractionDate"], errors="coerce")

print("Invalid SignupDate:", demographics_v2["SignupDate"].isna().sum())
print("Invalid TransactionDate:", transactions_v2["TransactionDate"].isna().sum())
print("Invalid InteractionDate:", social_v2["InteractionDate"].isna().sum())
```

```python
before_demo = demographics_v2.shape[0]
before_tx   = transactions_v2.shape[0]
before_sm   = social_v2.shape[0]

demographics_v2 = demographics_v2.dropna(subset=["SignupDate"])
transactions_v2 = transactions_v2.dropna(subset=["TransactionDate"])
social_v2       = social_v2.dropna(subset=["InteractionDate"])

print(f"Removed {before_demo - demographics_v2.shape[0]} rows from demographics")
print(f"Removed {before_tx - transactions_v2.shape[0]} rows from transactions")
print(f"Removed {before_sm - social_v2.shape[0]} rows from social media")

print("All datasets now have valid dates only")
```

## Customer ID's

```python
ids_demo = set(demographics_v2["CustomerID"])
tx_orphans = (~transactions_v2["CustomerID"].isin(ids_demo)).sum()
sm_orphans = (~social_v2["CustomerID"].isin(ids_demo)).sum()

print(f"Transactions without matching demographics: {tx_orphans}")
print(f"Social media interactions without matching demographics: {sm_orphans}")
```

```python
transactions_before = transactions_v2.shape[0]
social_before = social_v2.shape[0]

transactions_v2 = transactions_v2[transactions_v2["CustomerID"].isin(demographics_v2["CustomerID"])].copy()
social_v2 = social_v2[social_v2["CustomerID"].isin(demographics_v2["CustomerID"])].copy()

print(f"Dropped {transactions_before - transactions_v2.shape[0]} orphan transactions")
print(f"Dropped {social_before - social_v2.shape[0]} orphan social interactions")
print("All remaining records now have matching demographics")
```

## Standardization

```python
# --- Standardize Categories ---
def clean_str(x):
    if pd.isna(x): return x
    return str(x).strip().title()
```

```python
# demographics
demographics_v2["IncomeLevel"] = demographics_v2["IncomeLevel"].apply(clean_str)
demographics_v2["Gender"]      = demographics_v2["Gender"].apply(clean_str)
```

```python
# transactions
transactions_v2["ProductCategory"] = transactions_v2["ProductCategory"].apply(clean_str)
transactions_v2["PaymentMethod"]   = transactions_v2["PaymentMethod"].apply(clean_str)
```

```python
# social
social_v2["Platform"]  = social_v2["Platform"].apply(clean_str)
social_v2["Sentiment"] = social_v2["Sentiment"].apply(clean_str)
```

## Outlier

```python
print(transactions_v2["Amount"].describe())
```

```python
# Check top 5 largest transactions
print("\nTop 5 transaction amounts:")
print(transactions_v2.nlargest(5, "Amount")[["CustomerID","Amount","ProductCategory"]])
```

```python
amt_threshold = transactions_v2["Amount"].quantile(0.99)
transactions_v2["IsHighSpender"] = (transactions_v2["Amount"] > amt_threshold).astype(int)
print(f"Flagged {transactions_v2['IsHighSpender'].sum()} high-value transactions (>{amt_threshold:.2f})")
```

```python
# Remove negative transactions (likely refunds/errors)
neg_before = transactions_v2.shape[0]
transactions_v2 = transactions_v2[transactions_v2["Amount"] >= 0].copy()
print(f"Dropped {neg_before - transactions_v2.shape[0]} negative-amount transactions")
```

```python
# Keep high spenders but flagged
amt_threshold = transactions_v2["Amount"].quantile(0.99)
transactions_v2["IsHighSpender"] = (transactions_v2["Amount"] > amt_threshold).astype(int)
print(f"Flagged {transactions_v2['IsHighSpender'].sum()} high-value transactions (> {amt_threshold:.2f})")
```

## Conclusion

After cleaning the three datasets, we ensured all records are consistent, valid, and analysis-ready. From the demographics file (3,000 rows), we removed 98 entries with invalid signup dates, dropped duplicate customer IDs, and corrected unrealistic ages (e.g., -1, 150) by replacing them with the median, resulting in about 2,902 valid customer records.

In transactions (2,917 rows), 98 rows with invalid dates, 88 orphan records without demographics, and negative amounts were dropped, while 29 high-value transactions (1% above the 99th percentile) were flagged, leaving about 2,730 valid entries.


For social media (3,023 rows), 100 invalid interaction dates and 92 orphan records were removed, with missing platform and sentiment values replaced by “Unknown,” producing about 2,831 valid interactions.

Across all datasets, roughly 6–7% of rows were dropped to ensure integrity, all categorical nulls were standardized, and only realistic numeric values remain, leaving the data consistent and reliable for EDA and machine learning.
