```python
# Step 1: Import libraries
import pandas as pd

# Step 2: Load the contaminated dataset
file_name = "social_media_interactions_contaminated.csv"
df = pd.read_csv(file_name)

# Step 3: Check and remove duplicate rows
print("Duplicate rows before cleaning:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicate rows after cleaning:", df.duplicated().sum())

# Step 4: Identify columns with missing values
missing_values = df.isnull().sum()
print("\nColumns with missing values:")
print(missing_values[missing_values > 0])

# Step 5: Save the cleaned dataset as 'clean.csv'
output_file = "clean_social_media_interactions.csv"
df.to_csv(output_file, index=False)
print(f"\nCleaned dataset saved as {output_file}")
```
